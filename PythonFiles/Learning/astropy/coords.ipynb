```python
import numpy as np
import matplotlib.pyplot as plt
from astropy import units as u
from astropy.coordinates import SkyCoord, Distance
from astropy.table import QTable
from astroquery.gaia import Gaia

%matplotlib widget
```

```python
Gaia.ROW_LIMIT = 5000
ngc188_center = SkyCoord(12.11 * u.deg, 85.26 * u.deg, frame='icrs')
ngc188_center
```

```python
(ngc188_center.ra, ngc188_center.dec)
```

```python
ngc188_center.ra
```

```python
ngc188_center.dec
```

```python
job_async = Gaia.cone_search_async(ngc188_center, radius=0.5 * u.deg)
```

```python
ngc188_table = job_async.get_results()
```

```python
ngc188_table
```

```python
ngc188_table = ngc188_table[ngc188_table['phot_g_mean_mag'] < 19 * u.mag]
ngc188_table
```

```python
type(ngc188_table)
```

```python
ngc188_table['dec']
```

```python
ngc188_table = ngc188_table[ngc188_table['parallax'] / ngc188_table['parallax_error'] > 10]
ngc188_table
```

```python
dists = Distance(parallax=ngc188_table['parallax'].filled(np.nan))
dists
```

```python
coords = SkyCoord(
    ra=ngc188_table['ra'],
    dec=ngc188_table['dec'],
    distance=dists
)
coords
```

```python
fig = plt.figure()
ax = fig.add_subplot()
scatter = ax.scatter(
    coords.ra.degree,
    coords.dec.degree,
    c=coords.distance.kpc,
    s=5,
    cmap="twilight",
)

ax.set_xlabel("Right Ascension, deg")
ax.set_ylabel("Declination, deg")

fig.colorbar(scatter, orientation="vertical").set_label(f'Distance, {u.kpc:latex_inline}')

ax.set_title("Gaia DR2 sources near NGC 188", fontsize=18)

plt.show()
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
scatter = ax.scatter(
    coords.ra.degree,
    coords.distance.kpc,
    coords.dec.degree,
    c=coords.distance.kpc,
    s=5,
    cmap="twilight",
)

ax.set_xlabel("Right Ascension, deg")
ax.set_ylabel("Distance, kpc")
ax.set_zlabel("Declication, deg")

fig.colorbar(scatter, orientation="horizontal").set_label(f'Distance, {u.kpc:latex_inline}')

ax.set_title("Gaia DR2 sources near NGC 188", fontsize=18)

plt.show()
```
